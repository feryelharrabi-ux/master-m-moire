# Visualisation des résultats

Charge les métriques JSON produites par `train_baselines.py` ou `run_ai_eval.py` et trace les scores disponibles.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

results_path = Path('../data/parsed/baseline_metrics.json')
if not results_path.exists():
    raise FileNotFoundError(f'Générez d’abord les métriques : {results_path}')
results = json.loads(results_path.read_text(encoding='utf-8'))
results
```

```python
metrics = ['accuracy', 'precision', 'recall', 'f1']
models = list(results)
for metric in metrics:
    plt.figure(figsize=(7, 3))
    plt.bar(models, [results[name][metric] for name in models])
    plt.ylim(0, 1)
    plt.ylabel(metric)
    plt.title(f'{metric} par modèle')
    plt.xticks(rotation=15)
    plt.tight_layout()
    plt.show()
```
